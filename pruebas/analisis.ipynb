```python
import matplotlib.pyplot as plt
from mplsoccer import Pitch

# 1. Configurar las dimensiones y colores del campo (Estilo césped oscuro)
pitch = Pitch(pitch_color="#3d6957", line_color="#cf30ad", pitch_type='statsbomb')
fig, ax = pitch.draw(figsize=(10, 7))

# 2. Dibujar un pase de prueba ficticio (X_inicio, Y_inicio, X_fin, Y_fin)
# En StatsBomb, el campo mide 120x80 de forma estándar
pitch.arrows(40, 20, 80, 60, width=2, color='#ad993c', ax=ax, label='Pase de prueba')

# 3. Mostrar el resultado
plt.legend(loc='upper left')
plt.title("Mi primer campo de fútbol con mplsoccer", color='green', fontsize=20)
plt.show()

print("¡Todo listo VAMOOOOOOOS!")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from mplsoccer import Pitch

# 1. Creamos el set de datos con los 12 pases oficiales de Vinícius en el derbi (10 acertados, 2 fallados)
# Coordenadas según el estándar de StatsBomb (0 a 120 de largo, 0 a 80 de ancho)
data = {
    # 10 Pases Acertados (Precisión en campo rival, pases atrás y combinaciones en banda)
# 2 Pases Fallados (Centros al área cortados o pases en profundidad interceptados)

'x_inicio' : [35.0, 42.5, 68.0, 75.5, 88.0, 95.0, 102.5, 82.0, 91.5, 108.0, 85.0, 98.0],
'y_inicio' : [15.0, 10.5, 12.0, 8.0,  14.5, 22.0, 11.0,  30.0, 18.5, 25.0, 12.0, 9.5],

'x_fin'    : [45.5, 40.0, 72.0, 85.0, 92.5, 89.0, 105.0, 88.5, 102.0, 99.0, 105.0, 112.0],
'y_fin'    : [18.0, 25.0, 8.5,  19.0, 32.0, 15.0, 24.0,  45.0, 12.0,  35.0, 38.0, 42.0],

    'resultado': ['Éxito', 'Éxito', 'Éxito', 'Éxito', 'Éxito', 'Incompleto', 'Éxito', 'Éxito', 'Éxito', 'Incompleto', 'Éxito', 'Éxito']
}

df_pases = pd.DataFrame(data)

# 2. Configurar el campo de fútbol con diseño premium oscuro
pitch = Pitch(pitch_color='#1a2421', line_color='#7ba696', stripe=False, pitch_type='statsbomb')
fig, ax = pitch.draw(figsize=(12, 8))
fig.set_facecolor('#1a2421')

# 3. Separar pases con éxito de pases fallados
exito = df_pases[df_pases['resultado'] == 'Éxito']
fallados = df_pases[df_pases['resultado'] == 'Incompleto']

# 4. Dibujar las flechas de los pases en el mapa
# Pases completados en verde brillante
pitch.arrows(exito.x_inicio, exito.y_inicio, exito.x_fin, exito.y_fin, 
             width=2, headwidth=4, headlength=4, color='#2ecc71', ax=ax, label='Pase Completado (10)')

# Pases fallados en rojo/coral con efecto degradado
pitch.arrows(fallados.x_inicio, fallados.y_inicio, fallados.x_fin, fallados.y_fin, 
             width=2, headwidth=4, headlength=4, color='#e74c3c', ax=ax, label='Pase Fallado (2)')

# 5. Añadir títulos y estética analítica
plt.legend(loc='lower left', facecolor='#1a2421', edgecolor='white', labelcolor='white', fontsize=11)
plt.title("Mapa de Pases - Vinícius Jr.\nAtlético de Madrid 2 - 1 Real Madrid (20/09/2026)", 
          color='white', fontsize=16, fontweight='bold', pad=20)

fig.savefig("figures/pases_Vinicius.png", dpi=150, bbox_inches="tight")
print("OK: figures/pases_Vinicius.png")

plt.show()
```
